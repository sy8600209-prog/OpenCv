```python
# wroking with web cam or video
cap=cv2.VideoCapture(0)
while True:
    ret,frame=cap.read() # ret- true/false,frame=Image
    if not ret:
        print("Could not read frame")
        break
    cv2.imshow("webcam feed",frame)
    if cv2.waitKey(1)& 0xFF==ord('q'):# 1 msecond 30 to 60 images per second
         print("quitting")
         break
cap.release()
cv2.destroyAllWindows()
```

```python
# saving video
camera=cv2.VideoCapture(0)
frame_width=int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height=int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))
codec=cv2.VideoWriter_fourcc(*'XVID')
recorded=cv2.VideoWriter("my_video.avi",codec,20,(frame_width,frame_height))
while True:
    success,image=camera.read()
    if not success:
        break
    recorded.write(image)
    cv2.imshow("recording live",image)
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break

camera.release()
recorded.release()
cv2.destroyAllWindows()
```

```python
# image filtering
# remove noise
# smoth
# bakcgound blurr
image=cv2.imread("image2.jpg")
blurr=cv2.GaussianBlur(image,(7,7),0)# kernel (3,3)ligh blurr,sigma =0
cv2.imshow("original",image)
cv2.imshow("blurred",blurr)
cv2.waitKey(0)
cv2.destroyAllWindows()

# same as blurr=cv2.medianBlur(image,11)
```

```python
# filter
import numpy
sharpen_kernel=np.array([0,-1,0],[-1,5,-1,],[0,-1,0])
sharped=cv2.filter2D(image,-1,sharpen_kernel)
cv2.imshow("original",image)
cv2.imshow("blurred",sharped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Use OpenCV's built-in path and the correct human face cascade
cascade_path = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(cascade_path)

# Verify the cascade loaded correctly
if face_cascade.empty():
    print("Error: Could not load the face cascade file.")
    exit()

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        print("Error: Could not read frame from webcam.")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5
    )

    for (x, y, w, h) in faces:
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

    cv2.imshow("Webcam Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
